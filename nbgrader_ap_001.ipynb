**15.1 Eliminar  todo lo que se encuentre en la carpeta feedback (pero no la carpeta feedback)**

```python
import os
import shutil

# Ruta a la carpeta
folder_path = '/content/drive/MyDrive/nbgrader_PAGD1_14123/feedback/'

# Recorre todos los elementos dentro de la carpeta
for filename in os.listdir(folder_path):
    file_path = os.path.join(folder_path, filename)
    try:
        if os.path.isfile(file_path) or os.path.islink(file_path):
            os.unlink(file_path)
        elif os.path.isdir(file_path):
            shutil.rmtree(file_path)
    except Exception as e:
        print('Error al eliminar %s. Razón: %s' % (file_path, e))
```
